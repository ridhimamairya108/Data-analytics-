```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python

url = "https://raw.githubusercontent.com/46n/king-county-house-sales-analysis/main/data/housing.csv"

df = pd.read_csv(url)
```

```python
df.head()
```

```python

print("Dataset Shape:", df.shape)
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python

df.info()
```

```python

df.describe()
```

```python

df.isnull().sum()
```

```python

print("Duplicate rows:", df.duplicated().sum())
```

```python

data = df.copy()
```

```python

data = data.drop(columns=['id'], errors='ignore')
```

```python

print("Missing values before cleaning:")
print(data.isnull().sum())

for col in data.select_dtypes(include=np.number).columns:
    data[col] = data[col].fillna(data[col].median())

print("\nMissing values after cleaning:")
print(data.isnull().sum().sum())
```

```python

data = data.drop_duplicates()

print("Duplicates after cleaning:", data.duplicated().sum())
```

```python

if 'date' in data.columns:
    data['date'] = pd.to_datetime(data['date'])

print(data.dtypes)
```

```python

data['price_per_sqft'] = (
    data['price'] / data['sqft_living'].replace(0, np.nan)
)

data[['price', 'sqft_living', 'price_per_sqft']].head()
```

```python

if 'date' in data.columns:
    data['sale_year'] = data['date'].dt.year
    data['house_age'] = (
        data['sale_year'] - data['yr_built']
    )

data[['yr_built', 'sale_year', 'house_age']].head()

if 'date' in data.columns:
    data['sale_year'] = data['date'].dt.year
    data['house_age'] = (
        data['sale_year'] - data['yr_built']
    )

data[['yr_built', 'sale_year', 'house_age']].head()
```

```python

prices = data['price'].to_numpy()

print(prices[:10])
```

```python

print("Mean price:", np.mean(prices))
print("Median price:", np.median(prices))
```

```python

print("Minimum price:", np.min(prices))
print("Maximum price:", np.max(prices))
```

```python

print("Price standard deviation:", np.std(prices))
```

```python

print("25th percentile:", np.percentile(prices, 25))
print("50th percentile:", np.percentile(prices, 50))
print("75th percentile:", np.percentile(prices, 75))
```

```python

expensive_houses = prices[prices > np.median(prices)]

print("Houses above median price:", len(expensive_houses))
```

```python

price_in_thousands = prices / 1000

print(price_in_thousands[:10])
```

```python

sample_prices = np.random.default_rng(42).choice(
    prices, size=10, replace=False)

print("Random sample:", sample_prices)
```

```python

data['bedrooms'].value_counts().sort_index()
```

```python

data['price_category'] = pd.cut(
    data['price'],
    bins=[0, 300000, 600000, 1000000, float('inf')],
    labels=['Budget', 'Mid-range', 'Premium', 'Luxury']
)

data['price_category'].value_counts()
```

```python

data.select_dtypes(include=np.number).corr()['price'].sort_values(
    ascending=False
)
```

```python

if 'date' in data.columns:
    data['sale_month'] = data['date'].dt.month
    print(data.groupby('sale_month')['price'].mean())
```

```python

plt.figure(figsize=(7,4))

sns.histplot(data['price'], bins=40, kde=True)

plt.title("Distribution of House Prices")
plt.xlabel("House Price ($)")
plt.ylabel("Number of Houses")

plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(9,5))
sns.scatterplot(
    data=data,
    x='sqft_living',
    y='price',
    alpha=0.4
)
plt.title("House Price vs Living Area")
plt.xlabel("Living Area (sq ft)")
plt.ylabel("Price ($)")
plt.show()
```

```python

plt.figure(figsize=(9,5))
sns.boxplot(data=data, x='bedrooms', y='price')
plt.title("House Prices by Number of Bedrooms")
plt.xlabel("Bedrooms")
plt.ylabel("Price ($)")
plt.show()
```

```python

plt.figure(figsize=(13,9))
sns.heatmap(
    data.select_dtypes(include=np.number).corr(),
    cmap='coolwarm',
    annot=False
)
plt.title("Correlation Between Numerical Features")
plt.show()
```

```python

zip_prices = data.groupby('zipcode')['price'].median()

top_zip = zip_prices.sort_values(ascending=False).head(10)

plt.figure(figsize=(12,5))
top_zip.plot(kind='bar')
plt.title("Top 10 ZIP Codes by Median House Price")
plt.xlabel("ZIP Code")
plt.ylabel("Median Price ($)")
plt.xticks(rotation=45)
plt.show()
```

```python

plt.figure(figsize=(10,7))
plt.scatter(
    data['long'],
    data['lat'],
    c=data['price'],
    cmap='viridis',
    s=8,
    alpha=0.5
)
plt.colorbar(label='House Price ($)')
plt.title("Geographical Distribution of House Prices")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.show()
```
